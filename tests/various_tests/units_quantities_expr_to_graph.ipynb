# Testbed for non-integer powers of units and quantities
This Jupyter notebook is a testbed for the implementation of non-integer powers of units and of quantities in the SI Reference Point. It uses elements of the package SI-REFENCE-POINT.  
  
The function _transform_unit_expr_to_graph_ transforms an unit expression to a rdf graph (binary tree). It is defined in _units_abox.py_.
The function _transform_qty_expr_to_graph_ transforms an quantity expression to a rdf graph (binary tree). It is defined in _quantities_abox.py_.   
  
Both functions are structured in exactly the same way
 According to the description, the unit/quantity expression can be submitted as dict, list or strings
* **Dicts** will be exptected to be of the form {"mult": [{"exp": ["metre", 1]}, {"exp": ["second", -1]}]}
* **Lists** must be of the form [["(unit),(power)],["(unit),(power)]], e.g. [["meter",1],["second",-1]];
* **Strings** will be considered to represent a single unit, and turned into a URI;

This testbed is ment to check if the functions produce the correct outputs. 

  
created: 2026-07-17

```python
# import relevant modules
from si_ref_point.tboxes.si_tbox import SiElements
from si_ref_point.aboxes.units_abox import transform_unit_expr_to_graph as transform_unit_expr_to_graph
from si_ref_point.aboxes.quantities_abox import transform_qty_expr_to_graph as transform_qty_expr_to_graph
from rdflib import Graph, BNode,RDF, URIRef
from fractions import Fraction

si_graph = SiElements() 
```

#### Test of **unit** binary tree

```python
### binary tree builder for unit expression
empty_graph = Graph()
local_graph = Graph()

empty_graph.bind("units",si_graph.namespace_units)
empty_graph.bind("si",si_graph.namespace)

# unit expression to be transformed into a graph. 
# according to the description in the python script, three representations are possible:


#unit_expression = [["Newton",1],["meter",1]]
unit_expression =   [["Volt", 1], ["Hertz", "-1/2"]]
#unit_expression =   [["Volt", 1],["meter",-1]]


local_graph = transform_unit_expr_to_graph(unit_expression,si_graph,empty_graph)[0]


   
print(local_graph.serialize(format="ttl"))
```

#### Test of **quantity** binary tree

```python
### binary tree builder for unit expression
empty_graph = Graph()
local_graph = Graph()

empty_graph.bind("quantities",si_graph.namespace_quantities)
empty_graph.bind("si",si_graph.namespace)

# unit expression to be transformed into a graph. 
# according to the description in the python script, three representations are possible:


qty_expression = [["LENG",1],["TIME","-1/2"]]
#qty_expression =   [["Volt", 1], ["Hertz", "-1/2"]]
#qty_expression =   [["ELPD", 1],["LENG",-1]]


local_graph = transform_qty_expr_to_graph(qty_expression,si_graph,empty_graph)[0]

   
print(local_graph.serialize(format="ttl"))

```
